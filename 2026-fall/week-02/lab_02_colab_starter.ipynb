## 1. Get the file

This is the same `JEA Detail.txt` that is posted in the Week 2 module. Reading it straight from the web means there is no upload step and everyone starts from an identical file.

```python
import urllib.request

URL = "https://raw.githubusercontent.com/sean-mccaman/acctg5150-090/main/2026-fall/week-02/JEA%20Detail.txt"

raw = urllib.request.urlopen(URL).read()
print(f"downloaded {len(raw):,} bytes")

# Save a local copy so the rest of your work does not re-download it.
with open("JEA Detail.txt", "wb") as f:
    f.write(raw)
print("saved as 'JEA Detail.txt' in this Colab session")
```

## 2. Look at the raw bytes before you decode anything

Do not skip this. The first thing you need from this file is its **encoding**, and the bytes tell you directly.

Watch for two things in the output below:

- the first few bytes, which may be a byte order mark that names the encoding
- whether the letters are separated by `\x00` null bytes

If you see those nulls, the file is not the UTF-8 you were probably expecting, and any tool that assumes UTF-8 will fail on the very first read. The [encoding primer](https://seanmccaman.com/acctg6155/2026-fall/week-2/interactive/lab_02_encoding_primer.html) explains why.

```python
print("first 60 bytes, shown exactly as stored:")
print(raw[:60])
print()
print("the same bytes, one per column:")
print(" ".join(f"{b:02x}" for b in raw[:32]))
print()
print(f"null bytes in the first 1000: {raw[:1000].count(0)}")
```

## 3. Now read it as text

Once you know the encoding, decode with it and print the top of the file. What you are looking for is everything that is **not** data: the report header, the two-line column header, and the page-break rows scattered further down.

Fill in the encoding yourself. If you guess wrong you will get an error or nonsense, which is the point of step 2.

```python
ENCODING = ""  # <- fill this in from what you saw in step 2

if not ENCODING:
    print("Set ENCODING first. Step 2 tells you what it is.")
else:
    text = raw.decode(ENCODING)
    lines = text.splitlines()
    print(f"{len(lines):,} lines total\n")
    for i, line in enumerate(lines[:25]):
        print(f"{i:>3} | {line}")
```
