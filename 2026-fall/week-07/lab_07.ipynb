2. Run your own Week 3 aging query, or this one, which is the same query with the rows
   sorted in the same bucket order as your pandas table:

```sql
WITH paid AS (
  SELECT invoice_id, SUM(amount) AS total_paid FROM payments GROUP BY invoice_id
), balances AS (
  SELECT i.*, i.amount - COALESCE(p.total_paid, 0) AS balance
  FROM invoices AS i LEFT JOIN paid AS p ON p.invoice_id = i.invoice_id
), aged AS (
  SELECT *,
    CASE WHEN julianday('2026-04-30') - julianday(due_date) <= 0  THEN 'Current'
         WHEN julianday('2026-04-30') - julianday(due_date) <= 30 THEN '1-30'
         WHEN julianday('2026-04-30') - julianday(due_date) <= 60 THEN '31-60'
         WHEN julianday('2026-04-30') - julianday(due_date) <= 90 THEN '61-90'
         ELSE 'Over 90' END AS aging_bucket
  FROM balances WHERE balance > 0
)
SELECT aging_bucket, COUNT(*) AS invoice_count, ROUND(SUM(balance), 2) AS total_balance
FROM aged
GROUP BY aging_bucket
ORDER BY CASE aging_bucket WHEN 'Current' THEN 1 WHEN '1-30' THEN 2 WHEN '31-60' THEN 3
                           WHEN '61-90' THEN 4 ELSE 5 END;
```
